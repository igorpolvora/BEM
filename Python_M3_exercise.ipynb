1.2. Crie uma lista chamada `leads` com a renda dos usuários que clicaram na propaganda.

```
# Lista para armazenar as rendas dos usuários que clicaram na propaganda
leads = []

# Iterando sobre os dados dos usuários
for dado_de_usuario in propaganda_online:
    if dado_de_usuario.get('clicou_no_ad') == 1:  # Verifica se o usuário clicou no anúncio
        leads.append(dado_de_usuario['renda_area'])  # Adiciona a renda à lista leads

# Exibe a lista de rendas
print(leads)
```

1.3. Crie uma lista chamada `cidades` com a cidade dos usuários que passaram mais de 70 segundos no site.

```
# Lista para armazenar as cidades dos usuários que passaram mais de 70 segundos no site
cidades = []

# Iterando sobre os dados dos usuários
for dado_de_usuario in propaganda_online:
    if dado_de_usuario.get('tempo_gasto_site') and dado_de_usuario['tempo_gasto_site'] > 70:  # Verifica se o tempo no site é maior que 70
        cidades.append(dado_de_usuario['cidade'])  # Adiciona a cidade à lista cidades

# Exibe a lista de cidades
print(cidades)
```
